## Introduction

Operators are symbols or keywords that tell Python to perform an action on one or more values. They are used for calculations, comparing information, checking conditions, and making decisions in a program. Python has different groups of operators, and each group has its own purpose. In this assignment, I will explain the main types of Python operators with their syntax and simple examples. I will also discuss operator precedence, which determines the order in which Python evaluates an expression.

## 1. What Are Operators?

An operator performs an operation on values or variables. The values used with an operator are called operands. For example, in `10 + 5`, the plus sign (`+`) is the operator, while `10` and `5` are the operands.

Python operators are commonly grouped into the following categories:

- Arithmetic operators
- Assignment operators
- Comparison operators
- Logical operators
- Membership operators
- Identity operators
- Bitwise operators

Python also follows operator precedence rules to decide which operation should be performed first when an expression contains more than one operator.

## 2. Arithmetic Operators

Arithmetic operators are used to perform mathematical calculations such as addition, subtraction, multiplication, and division. They are useful in programs that work with prices, marks, quantities, measurements, and other numeric values.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `10 + 3` gives `13` |
| `-` | Subtraction | `10 - 3` gives `7` |
| `*` | Multiplication | `10 * 3` gives `30` |
| `/` | Division | `10 / 3` gives approximately `3.333` |
| `//` | Floor division | `10 // 3` gives `3` |
| `%` | Modulus (remainder) | `10 % 3` gives `1` |
| `**` | Exponentiation (power) | `2 ** 3` gives `8` |

**Basic syntax:** `result = first_value operator second_value`

```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
```

**Explanation:** The example uses two variables, `a` and `b`, to demonstrate the arithmetic operators. Normal division (`/`) returns a floating-point result, while floor division (`//`) rounds the quotient down to the next lower integer. The modulus operator (`%`) gives the remainder after division. The exponentiation operator (`**`) raises one number to the power of another.

## 3. Assignment Operators

Assignment operators are used to assign values to variables. The basic assignment operator is `=`, which stores the value on the right side in the variable on the left side. Python also has shorthand assignment operators that update a variable using its current value.

| Operator | Example | Equivalent operation |
|---|---|---|
| `=` | `x = 5` | Assigns `5` to `x` |
| `+=` | `x += 3` | `x = x + 3` |
| `-=` | `x -= 3` | `x = x - 3` |
| `*=` | `x *= 3` | `x = x * 3` |
| `/=` | `x /= 3` | `x = x / 3` |
| `//=` | `x //= 3` | `x = x // 3` |
| `%=` | `x %= 3` | `x = x % 3` |
| `**=` | `x **= 3` | `x = x ** 3` |
| `&=`, `|=`, `^=`, `>>=`, `<<=` | Bitwise updates | Apply the matching bitwise operation and assign the result |

**Basic syntax:** `variable operator= value`

```python
score = 50
print("Starting score:", score)

score += 10
print("After adding 10:", score)

score -= 5
print("After subtracting 5:", score)

score *= 2
print("After multiplying by 2:", score)

score /= 5
print("After dividing by 5:", score)
```

**Explanation:** The variable `score` starts at `50`. Each shorthand assignment operator changes its current value. For example, `score += 10` adds 10 to the existing score instead of creating a separate calculation. These operators make code shorter and easier to read.

## 4. Comparison Operators

Comparison operators compare two values. The result of a comparison is always a Boolean value: `True` or `False`. They are often used in `if` statements and loops to help a program make decisions.

| Operator | Meaning | Example |
|---|---|---|
| `==` | Equal to | `5 == 5` is `True` |
| `!=` | Not equal to | `5 != 3` is `True` |
| `>` | Greater than | `5 > 3` is `True` |
| `<` | Less than | `3 < 5` is `True` |
| `>=` | Greater than or equal to | `5 >= 5` is `True` |
| `<=` | Less than or equal to | `3 <= 5` is `True` |

**Basic syntax:** `first_value comparison_operator second_value`

```python
first_number = 15
second_number = 10

print("Equal:", first_number == second_number)
print("Not equal:", first_number != second_number)
print("Greater than:", first_number > second_number)
print("Less than:", first_number < second_number)
print("Greater than or equal:", first_number >= second_number)
print("Less than or equal:", first_number <= second_number)
```

**Explanation:** Python compares `first_number` and `second_number` in each statement. Since 15 is greater than 10, the greater-than comparison returns `True`, while the less-than comparison returns `False`. Comparison operators are useful for checking marks, age limits, passwords, and other conditions.

## 5. Logical Operators

Logical operators are used to combine conditions or reverse the result of a condition. They are commonly used when a program needs to check more than one requirement.

| Operator | Meaning | Example |
|---|---|---|
| `and` | True only when both conditions are true | `age >= 18 and has_id` |
| `or` | True when at least one condition is true | `is_weekend or is_holiday` |
| `not` | Reverses the Boolean result | `not is_logged_in` |

**Basic syntax:** `condition1 and condition2`, `condition1 or condition2`, or `not condition`.

```python
age = 20
has_student_id = True
is_weekend = False

print("Can enter the student area:", age >= 18 and has_student_id)
print("Can relax today:", is_weekend or has_student_id)
print("Is not the weekend:", not is_weekend)
```

**Explanation:** The `and` operator returns `True` only when both conditions are true. The `or` operator returns `True` when at least one condition is true. The `not` operator changes `True` to `False` and `False` to `True`. Python also uses short-circuit evaluation: with `and`, it may stop after finding a false condition, and with `or`, it may stop after finding a true condition.

## 6. Membership Operators

Membership operators check whether a value is present in a sequence or collection, such as a string, list, or tuple. They are useful when searching for an item or checking whether a word or character exists in text.

| Operator | Meaning | Example |
|---|---|---|
| `in` | Returns `True` if the value is present | `"a" in "apple"` |
| `not in` | Returns `True` if the value is not present | `"z" not in "apple"` |

**Basic syntax:** `value in collection` or `value not in collection`.

```python
subjects = ["Python", "Networking", "Database", "Mathematics"]

print("Python is in the list:", "Python" in subjects)
print("English is in the list:", "English" in subjects)
print("English is not in the list:", "English" not in subjects)

message = "Learning Python is interesting"
print("Python is in the message:", "Python" in message)
```

**Explanation:** The first membership check returns `True` because `"Python"` is an item in the `subjects` list. The second returns `False` because `"English"` is not in the list. Membership checks in strings are case-sensitive, so `"python"` and `"Python"` are treated as different text.

## 7. Identity Operators

Identity operators check whether two variables refer to the exact same object in memory. They are different from comparison operators, which check whether values are equal.

| Operator | Meaning |
|---|---|
| `is` | Returns `True` if both variables refer to the same object |
| `is not` | Returns `True` if they refer to different objects |

**Basic syntax:** `object1 is object2` or `object1 is not object2`.

For checking whether a variable has no value, use `value is None`. Do not use `is` as a replacement for `==` when comparing ordinary values.

```python
list_one = [1, 2, 3]
list_two = [1, 2, 3]
list_three = list_one

print("list_one == list_two:", list_one == list_two)
print("list_one is list_two:", list_one is list_two)
print("list_one is list_three:", list_one is list_three)
print("list_one is not list_two:", list_one is not list_two)
```

**Explanation:** `list_one` and `list_two` contain the same values, so `==` returns `True`. However, they are two separate list objects, so `is` returns `False`. `list_three` refers to the same object as `list_one`, so `list_one is list_three` returns `True`. This example shows the difference between checking equality and checking identity.

## 8. Bitwise Operators

Bitwise operators perform operations on the individual bits of integer values. Computers represent integers in binary form, using only `0` and `1`. These operators are often used in low-level programming, networking, permissions, and situations where individual bits need to be checked or changed.

| Operator | Name | Meaning |
|---|---|---|
| `&` | AND | A bit is 1 only if both bits are 1 |
| `|` | OR | A bit is 1 if at least one bit is 1 |
| `^` | XOR | A bit is 1 if the two bits are different |
| `~` | NOT | Flips the bits; in Python, `~x` equals `-(x + 1)` for integers |
| `<<` | Left shift | Shifts bits to the left |
| `>>` | Right shift | Shifts bits to the right |

**Basic syntax:** `number1 & number2`, `number1 | number2`, `number1 ^ number2`, `~number`, `number << positions`, or `number >> positions`.

```python
a = 5   # Binary: 0101
b = 3   # Binary: 0011

print("Bitwise AND:", a & b)
print("Bitwise OR:", a | b)
print("Bitwise XOR:", a ^ b)
print("Bitwise NOT of a:", ~a)
print("Left shift a by 1:", a << 1)
print("Right shift a by 1:", a >> 1)
```

**Explanation using binary values:**

The number 5 is `0101` in binary, and the number 3 is `0011` (shown here using four bits for easier understanding).

- AND (`&`): `0101 & 0011` gives `0001`, which is 1.
- OR (`|`): `0101 | 0011` gives `0111`, which is 7.
- XOR (`^`): `0101 ^ 0011` gives `0110`, which is 6.
- NOT (`~`): `~5` gives `-6` in Python because Python integers use signed, arbitrary-precision representation.
- Left shift (`<<`): `5 << 1` gives 10 because the bits move one place to the left.
- Right shift (`>>`): `5 >> 1` gives 2 because the bits move one place to the right.

For non-negative integers, shifting left by one position is similar to multiplying by 2, while shifting right by one position is similar to floor-dividing by 2.

## 9. Operator Precedence

Operator precedence determines the order in which Python performs operations in an expression. When an expression contains multiple operators, Python does not always calculate from left to right. It follows a set of precedence rules. Parentheses can be used to make the intended order clear and to control which operation happens first.

For example, multiplication has higher precedence than addition. Therefore, Python calculates `4 + 3 * 2` as `4 + 6`, which gives `10`. If we write `(4 + 3) * 2`, the parentheses make Python add first, giving `14`.

```python
answer_one = 4 + 3 * 2
answer_two = (4 + 3) * 2

print("Without parentheses:", answer_one)
print("With parentheses:", answer_two)
```

### Common Python Operator Precedence (High to Low)

The table below gives a simplified guide to common Python operators, from higher precedence to lower precedence. Operators on the same level generally follow their associativity rules. Exponentiation (`**`) is right-associative, while most other binary arithmetic operators are left-associative.

| Priority | Operators | Description |
|---:|---|---|
| 1 | `()` | Parentheses / grouping |
| 2 | `**` | Exponentiation |
| 3 | `+x`, `-x`, `~x` | Unary plus, minus, and bitwise NOT |
| 4 | `*`, `/`, `//`, `%`, `@` | Multiplication, division, floor division, modulus, matrix multiplication |
| 5 | `+`, `-` | Addition and subtraction |
| 6 | `<<`, `>>` | Bit shifts |
| 7 | `&` | Bitwise AND |
| 8 | `^` | Bitwise XOR |
| 9 | `|` | Bitwise OR |
| 10 | `<`, `<=`, `>`, `>=`, `!=`, `==`, `in`, `not in`, `is`, `is not` | Comparisons, membership, and identity |
| 11 | `not` | Logical NOT |
| 12 | `and` | Logical AND |
| 13 | `or` | Logical OR |

This is a beginner-friendly summary of common operators, not the complete Python language reference. Assignment expressions and conditional expressions have additional rules.

```python
# Multiplication is performed before addition
result_one = 10 + 4 * 2

result_two = (10 + 4) * 2

result_three = 2 + 3 ** 2 * 2

print("10 + 4 * 2 =", result_one)
print("(10 + 4) * 2 =", result_two)
print("2 + 3 ** 2 * 2 =", result_three)
```

**Explanation:** In the first expression, Python multiplies 4 by 2 before adding 10, so the result is 18. In the second expression, parentheses make Python add 10 and 4 first, then multiply the result by 2, giving 28. In the third expression, Python calculates the power first, then multiplication, and finally addition, giving 20. Using parentheses can make complicated expressions easier to understand.

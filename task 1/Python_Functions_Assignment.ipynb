## Introduction

Functions are an important part of Python programming. A function is a reusable block of code that performs a particular task. Instead of writing the same instructions many times, we can create a function and call it whenever we need it. Functions help make programs more organized, readable, and easier to maintain. In this assignment, I will explain how to define and use functions, pass information to them, return results, and work with different types of arguments and variable scope.

## 1. Defining a Function Using `def`

The `def` keyword is used to create a function in Python. A function contains a set of instructions that only run when the function is called. A function can be created once and used many times in a program.

**Syntax:**

```python
def function_name():
    statements

function_name()
```

- `def` tells Python that we are defining a function.
- `function_name` is the name given to the function.
- Parentheses `()` are used after the function name.
- The function body is indented.
- Calling the function by its name runs the code inside it.

````python
# Example: Define and call a simple function
def greet():
    print("Hello! Welcome to Python programming.")

greet()
````

In this example, `greet()` is the function name. The message is displayed when we call the function. If we do not call it, the code inside the function will not run.

## 2. Parameters and Arguments

Parameters and arguments are used to pass information into a function.

- **Parameter:** A variable written inside the parentheses when defining a function.
- **Argument:** The actual value supplied when calling the function.

**Syntax:**

```python
def function_name(parameter):
    statements

function_name(argument)
```

````python
# Example: Function with one parameter
def greet_student(name):
    print("Hello,", name)

greet_student("Aarav")
greet_student("Sita")
````

Here, `name` is the parameter, while `"Aarav"` and `"Sita"` are arguments. The function can use different argument values each time it is called.

### Multiple Parameters and Arguments

A function can have more than one parameter. When calling it, we provide the matching arguments in the same order, unless we use keyword arguments.

````python
# Example: Function with two parameters
def display_student(name, course):
    print(name, "is studying", course)

display_student("Nabin", "Information Technology")
````

In this example, the function receives the student's name and course, then displays both values.

## 3. The `return` Statement

The `return` statement sends a result from a function back to the part of the program where the function was called. It is useful when we want a function to calculate something and give the answer back for later use.

**Syntax:**

```python
def function_name(parameters):
    result = calculation
    return result

answer = function_name(arguments)
```

````python
# Example: Return the sum of two numbers
def add_numbers(a, b):
    total = a + b
    return total

result = add_numbers(10, 20)
print("The sum is:", result)
````

The function adds the two values and returns the result. We store that returned value in the variable `result` and then print it. A function without a written `return` statement returns `None` by default.

## 4. Default Arguments

A default argument is a parameter that already has a value in the function definition. If we do not provide a value for that parameter when calling the function, Python uses the default value. If we provide a different value, the new value is used instead.

**Syntax:**

```python
def function_name(parameter=default_value):
    statements
```

````python
# Example: Function with a default argument
def greet(name, message="Welcome"):
    print(message, name)

greet("Aarav")
greet("Sita", "Good morning")
````

In the first call, Python uses the default message `"Welcome"`. In the second call, we provide `"Good morning"`, so it replaces the default value.

## 5. Keyword Arguments

Keyword arguments are passed by writing the parameter name followed by an equal sign and its value. They make function calls easier to understand and allow us to provide arguments in a different order from the function definition.

**Syntax:**

```python
function_name(parameter1=value1, parameter2=value2)
```

````python
# Example: Calling a function with keyword arguments
def student_details(name, age, course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)

student_details(course="IT", name="Nabin", age=19)
````

Even though the arguments are written in a different order, Python matches each value with the parameter name. This makes the function call clear and readable.

## 6. The `*args` Parameter

`*args` allows a function to accept any number of positional arguments. Python collects the extra positional arguments into a tuple. The name `args` is a common convention, but the asterisk `*` is what gives it this behaviour.

**Syntax:**

```python
def function_name(*args):
    for item in args:
        statements
```

````python
# Example: Add any number of values
def add_all(*args):
    total = 0
    for number in args:
        total += number
    return total

print(add_all(10, 20))
print(add_all(5, 10, 15, 20))
````

The function can receive two, four, or more numbers without changing its definition. Inside the function, `args` behaves like a tuple containing all the supplied values.

## 7. The `**kwargs` Parameter

`**kwargs` allows a function to accept any number of keyword arguments. Python collects these arguments into a dictionary, where each parameter name becomes a key and its value becomes the dictionary value.

**Syntax:**

```python
def function_name(**kwargs):
    for key, value in kwargs.items():
        statements
```

````python
# Example: Display different student details
def display_details(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

display_details(name="Sita", age=20, course="IT", city="Kathmandu")
````

In this example, the function receives several named values. They are stored in `kwargs` as a dictionary, and the loop displays each key and value.

## 8. Local and Global Scope

Scope means the part of a program where a variable can be accessed.

### Local Scope

A local variable is created inside a function. It can normally be used only inside that function. It is not directly available outside the function.

### Global Scope

A global variable is created outside all functions. It can be accessed inside functions as well as in the main part of the program.

**Using the `global` keyword:** If we need to change the value of a global variable inside a function, we use the `global` keyword to tell Python that we mean the global variable, not a new local variable.

````python
# Example: Local and global variables
college = "Model Institute of Technology"  # Global variable

def show_information():
    student = "Aarav"  # Local variable
    print("Student:", student)
    print("College:", college)

show_information()
print("College outside the function:", college)
````

The variable `college` is global, so it can be accessed inside and outside the function. The variable `student` is local to `show_information()` and cannot normally be accessed outside that function.

````python
# Example: Changing a global variable
score = 50

def update_score():
    global score
    score = 80

update_score()
print("Updated score:", score)
````

The `global` keyword allows the function to update the global variable `score`. Without it, assigning to `score` inside the function would create a local variable instead.

## 9. Lambda Functions

A lambda function is a small anonymous function. It is written in one line and does not need a regular function name. Lambda functions can accept multiple arguments, but they contain only one expression. The result of that expression is returned automatically.

**Syntax:**

```python
lambda arguments: expression
```

Lambda functions are useful for short operations, especially when a small function is needed for a short time.

````python
# Example: Lambda function to multiply two numbers
multiply = lambda a, b: a * b

print(multiply(4, 5))
````

Here, the lambda function takes two arguments, `a` and `b`, multiplies them, and returns the result. It is similar to writing a short function using `def`.

````python
# Example: Use lambda with sorted()
students = [("Aarav", 82), ("Sita", 95), ("Nabin", 76)]

students_sorted = sorted(students, key=lambda student: student[1])

print(students_sorted)
````

In this example, `lambda student: student[1]` tells `sorted()` to arrange the students according to the marks stored at index 1 in each tuple.

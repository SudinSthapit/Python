## 1. Python Introduction

Python code is generally easy to read because it uses clear words and fewer symbols than many other programming languages.

A Python program is normally saved in a file with the `.py` extension. We can also write and run Python code in a Jupyter Notebook, where explanations and code are kept together. The examples in this assignment use simple statements to demonstrate the basic features of the language.

````python
print("Hello, Python!")
print("This is my first Python example.")
````

**Explanation:** The `print()` function displays information on the screen. In this example, Python displays two lines of text. Text must be written inside quotation marks.

## 2. Indentation

Indentation means leaving spaces at the beginning of a line of code. In Python, indentation is important because it shows which statements belong together in a block. Other languages may use curly brackets to group statements, but Python normally uses indentation for this purpose. Four spaces are commonly used for each indentation level. Code in the same block should use consistent indentation.

````python
age = 19

if age >= 18:
    print("You are an adult.")
    print("You can continue with the program.")

print("This line is outside the if block.")
````

**Explanation:** The two indented `print()` statements belong to the `if` block. They run only when the condition `age >= 18` is true. The final statement is not indented, so it is outside the block and runs after the `if` statement. Incorrect indentation can cause an `IndentationError` or make a program behave differently than expected.

## 3. Comments

Comments are notes written in a program to explain what the code does. Python ignores comments when it runs the program. Comments are useful because they make code easier for the programmer and other people to understand. A single-line comment starts with the hash symbol (`#`).

````python
# Store a student's name in a variable
student_name = "Alex"

# Display the student's name
print(student_name)
````

**Explanation:** The lines beginning with `#` are comments. They explain the purpose of the statements below them, but they are not displayed as program output. Comments should be short and useful rather than repeating every line of code.

## 4. The `print()` Function

The `print()` function is used to show text, numbers, variable values, and the results of expressions. It is one of the first functions beginners learn because it helps us see what a program is doing.

**Basic syntax:**

```python
print(value)
```

We can print more than one value by separating the values with commas. By default, Python places a space between values and moves to a new line after each `print()` call.

````python
print("My name is Sam.")
print(25)

course = "Information Technology"
semester = 1
print("Course:", course)
print("Semester:", semester)

print("Five plus three equals", 5 + 3)
````

**Explanation:** The first two statements display a sentence and a number. The next statements display variable values along with labels. The final statement calculates `5 + 3` and displays the result. This shows that `print()` can display both stored information and the result of an expression.

## 5. The `input()` Function

The `input()` function allows a program to receive information from the user through the keyboard. It displays a prompt and waits for the user to type something. The value returned by `input()` is a string, even if the user enters digits. If we need to perform calculations with a number entered by the user, we should convert it using functions such as `int()` for whole numbers or `float()` for decimal numbers.

**Basic syntax:**

```python
variable = input("Enter something: ")
```

````python
name = input("Enter your name: ")
print("Hello,", name)
````

**Explanation:** When this cell runs, it asks the user to enter a name. Python stores the entered text in the variable `name`, and then displays a greeting. Try running the cell and entering your own name.

````python
age = int(input("Enter your age: "))
next_year_age = age + 1
print("Next year, you will be", next_year_age)
````

**Explanation:** Here, `input()` receives the age as text. The `int()` function converts that text into a whole number so Python can add 1 to it. If the user enters something that is not a valid whole number, Python will show a `ValueError`.

## 6. Variables

A variable is a name used to store a value in a program. We can use variables to keep information and use it again later. In Python, we create a variable by writing its name, followed by an equal sign (`=`), and then the value. Python automatically identifies the data type in many basic cases, so we do not need to declare the type before assigning a value.

**Basic syntax:**

```python
variable_name = value
```

````python
student_name = "Riya"      # String (text)
student_age = 20           # Integer (whole number)
height = 1.68              # Float (decimal number)
is_student = True          # Boolean (True or False)

print(student_name)
print(student_age)
print(height)
print(is_student)
````

**Explanation:** This example uses four variables with different types of values. `student_name` stores text, `student_age` stores a whole number, `height` stores a decimal number, and `is_student` stores a Boolean value. A variable's value can also be changed later in the program.

````python
score = 70
print("Original score:", score)

score = 85
print("Updated score:", score)
````

**Explanation:** The variable `score` first stores the value `70`. Later, it is assigned a new value, `85`. When we print it again, Python displays the updated value. This is why variables are useful for storing information that may change while a program runs.

## 7. Variable Naming Rules

Variable names should describe the information they store. Python has rules that must be followed when creating variable names:

1. A variable name must begin with a letter (`A–Z` or `a–z`) or an underscore (`_`).
2. It cannot begin with a number.
3. It can contain letters, numbers, and underscores, but it cannot contain spaces or symbols such as `@`, `-`, or `!`.
4. Variable names are case-sensitive. For example, `name` and `Name` are treated as different variables.
5. Python keywords, such as `if`, `for`, and `class`, cannot be used as variable names.
6. It is good practice to use meaningful names. For example, `student_name` is clearer than `x` when storing a student's name.

Python commonly uses **snake_case** for variable names. In this style, words are written in lowercase and separated by underscores.

````python
# Valid variable names
student_name = "Aarav"
total_marks = 450
_attendance = 90
subject2 = "Python"

print(student_name)
print(total_marks)
print(_attendance)
print(subject2)
````

**Examples of invalid variable names:**

| Invalid name | Reason |
|---|---|
| `2name` | Starts with a number |
| `student name` | Contains a space |
| `total-marks` | Contains a hyphen |
| `class` | Uses a Python keyword |

The following code is an example of correct naming. Invalid names are shown above as text because running them as Python assignments would cause syntax errors.

````python
name = "Maya"
Name = "Sita"

print(name)
print(Name)
````

**Explanation:** The output is different because Python treats uppercase and lowercase letters as different characters. Therefore, `name` and `Name` are two separate variables. Using clear and consistent names makes a program easier to read and maintain.

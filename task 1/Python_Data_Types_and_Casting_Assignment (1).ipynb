# Python Data Types and Type Casting

Topic: Data Types and Type Conversion in Python  

## Introduction

In Python, data types describe the kind of value a variable stores. For example, a person's age is usually stored as an integer, while a height can be stored as a floating-point number. Python also has data types for text, Boolean values, and values that represent the absence of a value. Understanding data types is important because it helps us choose the correct operations and avoid errors in a program. In this assignment, I will explain the main basic data types in Python and demonstrate how to check and convert them.

## 1. What Are Data Types?

A data type tells Python what kind of data a value represents. Different types are used for different purposes. For example, we can use numbers for calculations, strings for names and messages, and Boolean values for conditions. Python automatically assigns a data type when we give a value to a variable.

The main basic data types covered in this assignment are:

- `int` – stores whole numbers.
- `float` – stores numbers with decimal points.
- `str` – stores text.
- `bool` – stores either `True` or `False`.
- `NoneType` – represents the absence of a value, using `None`.

## 2. Integer (`int`)

An integer is a whole number that does not have a decimal part. Integers can be positive, negative, or zero. They are commonly used for values such as age, quantity, number of students, and scores when decimal values are not needed.

**Example syntax:**

```python
age = 20
number_of_students = 35
temperature_change = -4
```

````python
age = 20
number_of_students = 35
temperature_change = -4

print("Age:", age)
print("Number of students:", number_of_students)
print("Temperature change:", temperature_change)
````

**Explanation:** All three variables store whole numbers, so Python identifies their type as `int`. Integers are useful when we need to count items or work with values that do not require decimal places.

## 3. Floating-Point Number (`float`)

A float is a number that contains a decimal point. It is useful when a value needs more detail than a whole number can provide. Examples include height, weight, price, and percentage. Even a number such as `5.0` is treated as a float because it is written with a decimal point.

**Example syntax:**

```python
height = 1.75
price = 99.50
```

````python
height = 1.75
price = 99.50
average_marks = 82.5

print("Height:", height)
print("Price:", price)
print("Average marks:", average_marks)
````

**Explanation:** The values in this example contain decimal points, so their data type is `float`. Floats are useful for calculations where decimal values are important. However, some decimal calculations can have small precision differences because of how computers store floating-point numbers.

## 4. String (`str`)

A string is used to store text. It can contain letters, numbers, spaces, and special characters. A string is written inside single quotation marks or double quotation marks. Even if a string contains digits, Python treats it as text when it is inside quotation marks.

**Example syntax:**

```python
student_name = "Alex"
course = 'Information Technology'
phone_number = "9800000000"
```

````python
student_name = "Alex"
course = 'Information Technology'
phone_number = "9800000000"

print("Student name:", student_name)
print("Course:", course)
print("Phone number:", phone_number)
````

**Explanation:** All three values are strings because they are written inside quotation marks. A phone number is often stored as a string because we do not normally perform mathematical calculations with it, and it may contain leading zeros or symbols.

## 5. Boolean (`bool`)

A Boolean data type has only two possible values: `True` or `False`. It is mainly used in conditions and decision-making statements. For example, a program can use a Boolean value to check whether a student has passed or whether a user is logged in. In Python, the first letter of `True` and `False` must be capitalized.

**Example syntax:**

```python
is_logged_in = True
has_passed = False
```

````python
is_logged_in = True
has_passed = False

print("Logged in:", is_logged_in)
print("Passed:", has_passed)

if is_logged_in:
    print("Welcome back!")
else:
    print("Please log in.")
````

**Explanation:** The variables store Boolean values. The `if` statement checks whether `is_logged_in` is `True`. Since it is true in this example, the program displays the welcome message. Boolean values are useful whenever a program needs to make a decision based on a condition.

## 6. None (`NoneType`)

`None` is a special value in Python that represents no value or the absence of a value. It is different from zero, an empty string, and `False`. Python uses the data type name `NoneType` for this value. We can assign `None` to a variable when a value has not been provided yet or when we want to represent an empty result.

**Example syntax:**

```python
result = None
```

````python
result = None
middle_name = None

print("Result:", result)
print("Middle name:", middle_name)
print("Is the result missing?", result is None)
````

**Explanation:** Both variables currently contain `None`, which means no value has been assigned to them. The expression `result is None` checks whether the variable contains the special `None` value. It is common practice to use `is None` when checking for `None` in Python.

## 7. Checking Data Types with `type()`

The `type()` function is used to find the data type of a value or variable. It is helpful when learning Python because it allows us to check how Python understands the value we have stored. The function returns the type of the given object.

**Basic syntax:**

```python
type(value)
```

````python
age = 21
height = 1.72
name = "Maya"
is_student = True
result = None

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
print(type(result))
````

**Explanation:** The output shows the type of each variable. Python identifies `age` as an integer, `height` as a float, `name` as a string, `is_student` as a Boolean, and `result` as `NoneType`. The `type()` function is especially useful when checking input values or debugging a program.

## 8. Type Casting and Type Conversion

Type casting, also called type conversion, means changing a value from one data type to another. This is useful when a program receives data in one form but needs to use it in another. For example, the `input()` function returns text, so we often convert the entered value into an integer or float before doing calculations.

Python provides built-in conversion functions such as:

- `int()` – converts a suitable value into an integer.
- `float()` – converts a suitable value into a floating-point number.
- `str()` – converts a value into a string.
- `bool()` – converts a value into a Boolean value.

Not every conversion is possible. If a value cannot be converted into the requested type, Python may raise an error such as `ValueError` or `TypeError`.

### 8.1 Converting to Integer with `int()`

The `int()` function converts a suitable value into a whole number. When a float is converted to an integer, the decimal part is removed; Python does not round it to the nearest whole number. A string can also be converted if it contains a valid integer representation.

**Syntax:**

```python
int(value)
```

````python
number_from_text = int("25")
number_from_float = int(8.9)

print(number_from_text)
print(number_from_float)
print(type(number_from_text))
print(type(number_from_float))
````

**Explanation:** The text `"25"` becomes the integer `25`. The float `8.9` becomes the integer `8` because `int()` removes the decimal part instead of rounding the value. A string such as `"8.9"` cannot be converted directly with `int()` because it contains a decimal point.

### 8.2 Converting to Float with `float()`

The `float()` function converts a suitable value into a floating-point number. It can convert integers and strings containing valid decimal or whole-number text into floats.

**Syntax:**

```python
float(value)
```

````python
price_from_text = float("99.50")
number_from_integer = float(10)

print(price_from_text)
print(number_from_integer)
print(type(price_from_text))
print(type(number_from_integer))
````

**Explanation:** The string `"99.50"` is converted into a float, and the integer `10` becomes `10.0`. This conversion is useful when working with measurements, prices, averages, and other values that may contain decimal places.

### 8.3 Converting to String with `str()`

The `str()` function converts a value into text. It can convert numbers and Boolean values into strings. This is useful when we want to combine a value with a message or display it as text.

**Syntax:**

```python
str(value)
```

````python
age = 20
marks = 87.5

age_text = str(age)
marks_text = str(marks)

print("My age is " + age_text)
print("My marks are " + marks_text)
print(type(age_text))
print(type(marks_text))
````

**Explanation:** The integer `20` and float `87.5` are converted into strings. This allows them to be joined with other text using the `+` operator. Without converting the numbers first, joining a string and a number directly with `+` would cause a `TypeError`.

### 8.4 Converting to Boolean with `bool()`

The `bool()` function converts a value into either `True` or `False`. In general, values such as `0`, `0.0`, empty strings, and `None` are treated as `False`. Most other values, including non-empty strings and non-zero numbers, are treated as `True`.

**Syntax:**

```python
bool(value)
```

````python
print(bool(1))
print(bool(0))
print(bool("Hello"))
print(bool(""))
print(bool(None))
print(bool(5.5))
````

**Explanation:** The output shows how Python checks whether a value is considered true or false. A non-empty string such as `"Hello"` is `True`, while an empty string is `False`. One important point is that `bool("False")` is also `True` because the string contains text and is not empty. The word `False` inside quotation marks is not the same as the Boolean value `False`.

````python
print("bool('False'):", bool("False"))
print("bool(False):", bool(False))
print("bool(''):", bool(""))
````

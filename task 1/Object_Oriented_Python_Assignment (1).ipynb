# Object-Oriented Programming in Python



**Topic:** Classes and Objects, `self`, `__init__`, Methods, Variables, Inheritance, Polymorphism, Encapsulation, and MRO

## Introduction

Object-oriented programming (OOP) is a way of writing programs by grouping related data and behaviour together. In Python, a **class** acts like a blueprint, while an **object** is a particular instance created from that blueprint.

For example, a `Student` class can describe the information and actions common to students. Each student object can then have its own name, roll number, and marks.

This notebook explains the main OOP concepts with short examples. Run the code cells in order to see how each concept works.


## 1. Classes and Objects

A **class** defines the attributes and methods that its objects can use. An **object** is created by calling the class, much like calling a function.

### Basic syntax

```python
class ClassName:
    # attributes and methods
    pass

object_name = ClassName()
```


````python
class Student:
    def introduce(self):
        print("Hello, I am a student.")

student1 = Student()
student2 = Student()

student1.introduce()
student2.introduce()
````

In this example, `Student` is the class and `student1` and `student2` are two separate objects. Both objects can use the `introduce()` method.

**Key point:** A class describes what its objects can contain and do; each object is an individual instance of that class.


## 2. The `self` Parameter

The word `self` refers to the current object. It lets a method access that object's attributes and other methods. Python passes the object automatically when an instance method is called, so we normally do not pass `self` ourselves.

`self` is a convention rather than a reserved keyword, but it should be used for the first parameter of instance methods.

### Syntax

```python
def method_name(self):
    # work with this object
    pass
```


````python
class Student:
    def set_name(self, name):
        self.name = name

    def show_name(self):
        print("Student name:", self.name)

student1 = Student()
student2 = Student()

student1.set_name("Aarav")
student2.set_name("Sita")

student1.show_name()
student2.show_name()
````

The two objects store different values in `self.name`. When `student1.show_name()` runs, `self` refers to `student1`; when `student2.show_name()` runs, it refers to `student2`.


## 3. The `__init__()` Method

`__init__()` is an initializer method that Python calls after a new object has been created. It is commonly used to give the object's attributes their starting values.

### Syntax

```python
class ClassName:
    def __init__(self, parameter):
        self.attribute = parameter
```


````python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def display(self):
        print(f"Name: {self.name}")
        print(f"Roll number: {self.roll_no}")
        print(f"Marks: {self.marks}")

student = Student("Aarav", 12, 86)
student.display()
````

When `Student("Aarav", 12, 86)` is evaluated, Python creates an object and calls `__init__()` with the supplied values. The method stores them as attributes on that object.

**Remember:** `__init__()` initializes an object; it is not technically the method that creates the object.


## 4. Instance Methods

An **instance method** works with a particular object. Its first parameter is usually `self`, and it can read or change that object's instance variables.

### Syntax

```python
def method_name(self, other_parameters):
    # use self.attribute or call another method
    pass
```


````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited Rs. {amount}")
        else:
            print("Deposit amount must be positive.")

    def show_balance(self):
        print(f"{self.owner}'s balance: Rs. {self.balance}")

account = BankAccount("Sita", 1500)
account.deposit(500)
account.show_balance()
````

The `deposit()` and `show_balance()` methods operate on the account object. The deposit method also checks that the amount is positive before changing the balance.


## 5. Instance Variables and Class Variables

These variables differ in where their values belong:

- **Instance variable:** Belongs to one object. It is often created with `self`, usually inside `__init__()`.
- **Class variable:** Belongs to the class and is shared by instances unless an instance shadows it with its own attribute.

### Syntax

```python
class Example:
    category = "General"       # class variable

    def __init__(self, value):
        self.value = value     # instance variable
```


````python
class Student:
    school = "Everest Secondary School"  # class variable

    def __init__(self, name, roll_no):
        self.name = name                 # instance variables
        self.roll_no = roll_no

student1 = Student("Aarav", 12)
student2 = Student("Sita", 18)

print(student1.name, "-", student1.roll_no)
print(student2.name, "-", student2.roll_no)
print("School:", student1.school)
print("School:", student2.school)

Student.school = "Himalaya College"
print("Updated school:", student1.school)
print("Updated school:", student2.school)
````

Both students have their own `name` and `roll_no`, but they share the class variable `school`. Updating `Student.school` changes the value seen through both objects, as long as neither object has its own separate `school` attribute.


## 6. Static Methods

A **static method** is related to a class but does not automatically receive the instance (`self`) or the class (`cls`). It is marked with `@staticmethod` and is useful for a helper operation that logically belongs with the class.

### Syntax

```python
class ClassName:
    @staticmethod
    def helper(parameters):
        return result
```


````python
class Temperature:
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (celsius * 9 / 5) + 32

print(Temperature.celsius_to_fahrenheit(0))
print(Temperature.celsius_to_fahrenheit(25))
````

The conversion formula does not need information from a particular `Temperature` object, so a static method is suitable here. A static method can be called through the class without creating an instance.


## 7. Inheritance

**Inheritance** allows a new class to reuse and extend the attributes and methods of an existing class.

- **Parent/base class:** The class being inherited from.
- **Child/derived class:** The class that inherits from the parent.

### Syntax

```python
class Parent:
    pass

class Child(Parent):
    pass
```


````python
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f"My name is {self.name}.")

class Student(Person):
    def study(self):
        print(f"{self.name} is studying Python.")

student = Student("Aarav")
student.introduce()  # inherited from Person
student.study()      # defined in Student
````

`Student` inherits from `Person`, so it can use `introduce()` as well as its own `study()` method. This avoids repeating the same name-handling code in both classes.


## 8. Using `super()`

`super()` gives access to methods in the next class in the method resolution order (MRO). A common use is calling the parent class's `__init__()` from a child class so the inherited attributes are initialized.

### Syntax

```python
class Child(Parent):
    def __init__(self, value, extra):
        super().__init__(value)
        self.extra = extra
```


````python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, course):
        super().__init__(name)
        self.course = course

    def display(self):
        print(f"Name: {self.name}")
        print(f"Course: {self.course}")

student = Student("Sita", "Computer Science")
student.display()
````

`super().__init__(name)` calls the next initializer in the MRO, which in this example is `Person.__init__()`. The child class then adds its own `course` attribute.

In multiple inheritance, `super()` follows the MRO rather than simply meaning “call my direct parent.”


## 9. Method Overriding

**Method overriding** happens when a child class defines a method with the same name as a method in its parent class. When that method is called on a child object, Python uses the child's version.

### Syntax

```python
class Parent:
    def show(self):
        print("Parent version")

class Child(Parent):
    def show(self):
        print("Child version")
```


````python
class Animal:
    def sound(self):
        print("The animal makes a sound.")

class Dog(Animal):
    def sound(self):
        print("The dog barks.")

class Cat(Animal):
    def sound(self):
        print("The cat meows.")

Animal().sound()
Dog().sound()
Cat().sound()
````

`Dog` and `Cat` both override `sound()`. Each object responds using the implementation defined by its own class.


## 10. Polymorphism

**Polymorphism** means “many forms.” In OOP, different object types can be used through a common interface, while each type provides its own behaviour. Method overriding is one way to achieve this.

Python often uses **duck typing**: if an object supports the operation the code needs, the code can work with it without requiring a specific class.

### Example


````python
class Dog:
    def sound(self):
        return "Woof!"

class Cat:
    def sound(self):
        return "Meow!"

class Cow:
    def sound(self):
        return "Moo!"

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    print(animal.sound())
````

The loop does not need to check whether each object is a `Dog`, `Cat`, or `Cow`. It simply calls `sound()`, and each object supplies its own result. That is polymorphism in practice.


## 11. Encapsulation

**Encapsulation** means keeping data and the operations that work on it together, while controlling how that data is accessed or changed.

Python uses naming conventions rather than strict private-access enforcement:

- `name`: public attribute
- `_name`: intended for internal use by convention
- `__name`: name-mangled attribute, which helps avoid accidental access and clashes in subclasses; it is not completely inaccessible

A common approach is to use a property to validate changes to an attribute.

### Syntax

```python
class Example:
    def __init__(self, value):
        self._value = value

    @property
    def value(self):
        return self._value

    @value.setter
    def value(self, new_value):
        # validate new_value
        self._value = new_value
```


````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            print("Please enter a positive amount.")
            return
        self._balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            print("Please enter a positive amount.")
        elif amount > self._balance:
            print("Not enough money in the account.")
        else:
            self._balance -= amount

account = BankAccount("Aarav", 2000)
account.deposit(500)
account.withdraw(300)
print("Current balance:", account.balance)
account.withdraw(5000)
````

The balance is stored in `_balance`, which signals that it is intended for internal use. The `balance` property lets other code read the value, while the `deposit()` and `withdraw()` methods control updates and apply simple checks.

The underscore is not a security barrier: Python code can still access `_balance` directly. The design encourages callers to use the methods provided by the class.


## 12. Method Resolution Order (MRO)

The **Method Resolution Order** is the order Python follows to look for a method or attribute through a class's inheritance hierarchy. It matters especially when a class inherits from more than one parent.

Python uses the C3 linearization algorithm to create a consistent order. You can inspect the order with:

```python
ClassName.mro()
# or
ClassName.__mro__
```


````python
class A:
    def greet(self):
        print("Hello from A")

class B(A):
    def greet(self):
        print("Hello from B")

class C(A):
    def greet(self):
        print("Hello from C")

class D(B, C):
    pass

obj = D()
obj.greet()

print("MRO:")
for cls in D.mro():
    print(cls.__name__)
````

`D` inherits from `B` first and `C` second. Its MRO is `D`, `B`, `C`, `A`, and then `object`. Since both `B` and `C` define `greet()`, Python finds `B.greet()` first.

The order is not just a simple left-to-right search of every parent at every level; Python's MRO keeps the inheritance order consistent.

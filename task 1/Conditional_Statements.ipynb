# Python Conditional Statements

In a program, we often need to make decisions. For example, a program might check whether a student has passed an exam, whether someone is old enough to vote, or whether a number is positive or negative.

Python uses **conditional statements** to make these decisions. A condition is an expression that Python evaluates as either `True` or `False`. Depending on the result, the program runs a particular block of code.

In this notebook, we will learn about `if`, `elif`, and `else`, how to write nested conditions, and how to combine comparison and logical operators to make more useful decisions.

## 1. The `if` statement

The `if` statement runs a block of code only when its condition is `True`.

**Syntax:**

```python
if condition:
    statement
```

Notice the colon (`:`) at the end of the condition. The code that belongs to the `if` statement must be indented. Python commonly uses four spaces for indentation.



````python
age = 20

if age >= 18:
    print("You are eligible to vote.")
````

If the condition is `False`, Python skips the indented block and continues with the next statement.



````python
age = 15

if age >= 18:
    print("You are eligible to vote.")

print("Program finished.")
````

## 2. The `if-else` statement

Sometimes we want the program to choose between two possibilities. The `else` block runs when the `if` condition is `False`.

**Syntax:**

```python
if condition:
    statements_if_true
else:
    statements_if_false
```

Only one of these two blocks will run.



````python
number = 7

if number % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
````

Here is another example that checks whether a student has passed an exam. In this example, a score of 40 or above is considered a pass.



````python
marks = 35

if marks >= 40:
    print("Result: Pass")
else:
    print("Result: Fail")
````

## 3. The `if-elif-else` statement

When there are more than two possible outcomes, we can use `elif`. It is short for "else if" and lets us check another condition if the earlier condition was `False`.

Python checks the conditions from top to bottom. As soon as one condition is `True`, its block runs and the remaining conditions are skipped. The `else` block is optional and runs only when none of the conditions are `True`.

**Syntax:**

```python
if condition1:
    statements
elif condition2:
    statements
elif condition3:
    statements
else:
    statements
```

You can use as many `elif` blocks as you need.



````python
marks = 78

if marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 40:
    print("Grade: D")
else:
    print("Grade: F")
````

The order of conditions matters. For example, if you check `marks >= 40` first, a score of 85 will match that condition and Python will not check the higher grade conditions below it. It is usually best to check ranges from highest to lowest when using this kind of grading example.

## 4. Nested `if` statements

A nested `if` statement is an `if` statement placed inside another `if` statement. This is useful when one decision should only be made after another condition has been met.

**Syntax:**

```python
if condition1:
    if condition2:
        statements
    else:
        statements
else:
    statements
```

Each level must be indented correctly. The inner condition is checked only when the outer condition is `True`.



````python
age = 22
has_id = True

if age >= 18:
    print("Age requirement met.")

    if has_id:
        print("You can enter.")
    else:
        print("Please show your ID.")
else:
    print("You must be at least 18 to enter.")
````

In this example, the program first checks the age. Only if the person is at least 18 does it go on to check whether they have an ID.

### Another example: checking a username and password

A nested condition can also be used when a second check depends on the result of the first one.



````python
username = "student"
password = "python123"

if username == "student":
    if password == "python123":
        print("Login successful.")
    else:
        print("Incorrect password.")
else:
    print("Username not found.")
````

## 5. Comparison operators in conditions

Comparison operators are used to compare two values. The result of a comparison is always a Boolean value: `True` or `False`.

| Operator | Meaning | Example |
|---|---|---|
| `==` | Equal to | `5 == 5` gives `True` |
| `!=` | Not equal to | `5 != 3` gives `True` |
| `>` | Greater than | `8 > 5` gives `True` |
| `<` | Less than | `3 < 7` gives `True` |
| `>=` | Greater than or equal to | `5 >= 5` gives `True` |
| `<=` | Less than or equal to | `4 <= 6` gives `True` |

Remember that `=` is used for assignment, while `==` is used to compare two values.



````python
a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= 10)
print(b <= 5)
````

### Using comparison operators in an `if` statement



````python
temperature = 32

if temperature > 30:
    print("It is a hot day.")
else:
    print("The weather is not too hot.")
````

## 6. Logical operators

Logical operators allow us to combine conditions or reverse the result of a condition.

Python has three main logical operators:

| Operator | Meaning | When it is `True` |
|---|---|---|
| `and` | Logical AND | Both conditions are `True` |
| `or` | Logical OR | At least one condition is `True` |
| `not` | Logical NOT | The condition is `False` |

### `and` operator

The `and` operator returns `True` only when both conditions are `True`.



````python
age = 22
has_id = True

if age >= 18 and has_id:
    print("Entry allowed.")
else:
    print("Entry not allowed.")
````

### `or` operator

The `or` operator returns `True` when at least one of the conditions is `True`.



````python
day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("It is the weekend.")
else:
    print("It is a weekday.")
````

### `not` operator

The `not` operator reverses a Boolean result. It turns `True` into `False` and `False` into `True`.



````python
is_raining = False

if not is_raining:
    print("You can go for a walk.")
else:
    print("It is better to stay indoors.")
````

## 7. Combining comparison and logical operators

In real programs, decisions often depend on more than one piece of information. We can combine comparison operators with `and`, `or`, and `not` to write conditions that are more specific.

### Example 1: Checking eligibility

A student is eligible for a scholarship if their marks are at least 80 and their attendance is at least 75%.



````python
marks = 85
attendance = 80

if marks >= 80 and attendance >= 75:
    print("You are eligible for the scholarship.")
else:
    print("You are not eligible for the scholarship.")
````

### Example 2: Checking a discount

A shop gives a discount if the customer's total purchase is at least 5000 or the customer has a membership card.



````python
purchase_amount = 3500
has_membership = True

if purchase_amount >= 5000 or has_membership:
    print("You are eligible for a discount.")
else:
    print("No discount available.")
````

### Example 3: Checking an age range

Suppose an event is open to people aged from 18 to 30, including both 18 and 30.



````python
age = 25

if age >= 18 and age <= 30:
    print("You are within the allowed age range.")
else:
    print("You are outside the allowed age range.")
````

Python also allows a shorter way to write a range comparison:

`18 <= age <= 30`

This means the age must be at least 18 and at most 30.



````python
age = 17

if 18 <= age <= 30:
    print("Age accepted.")
else:
    print("Age not accepted.")
````

### Example 4: Combining `and`, `or`, and `not`

The following example checks whether a person can use a student discount. A person must be a student and either have a valid student card or be verified online.



````python
is_student = True
has_student_card = False
is_verified_online = True

if is_student and (has_student_card or is_verified_online):
    print("Student discount approved.")
else:
    print("Student discount not approved.")
````

Parentheses help make the intended grouping clear. Python evaluates the expression inside the parentheses first.

## 8. Truthy and falsy values

Conditions do not always have to be written as comparisons. Python can also evaluate many values directly as `True` or `False`.

Some common falsy values are:

- `False`
- `None`
- Numeric zero, such as `0` or `0.0`
- Empty strings, such as `""`
- Empty collections, such as `[]`, `()`, and `{}`

Most other values are considered truthy.

For example, this is a common way to check whether a list contains any items:



````python
items = ["pen", "book", "notebook"]

if items:
    print("The list contains items.")
else:
    print("The list is empty.")
````

And here is what happens with an empty list:



````python
items = []

if items:
    print("The list contains items.")
else:
    print("The list is empty.")
````

## 1. Tuples

A **tuple** is a built-in Python data type used to store more than one value in a single variable. A tuple is ordered, which means its items have a fixed position, and it allows duplicate values. The main feature of a tuple is that it is **immutable**, meaning its items cannot be changed after the tuple has been created.

Tuples are useful when a group of values should remain the same throughout a program. For example, a student's date of birth or the coordinates of a location can be stored in a tuple.

### Basic syntax

```python
tuple_name = (item1, item2, item3)
```

Example:


````python
student = ("Aarav", 19, "Information Technology")
print(student)
print(type(student))
````

In this example, the tuple named `student` stores a name, age, and course. The `type()` function confirms that the variable is a tuple.

**Important:** A tuple with only one item must include a comma. Without the comma, Python treats the value as an ordinary value inside parentheses.

````python
single_item = ("Python",)
print(single_item)
print(type(single_item))
````

## 2. Tuple packing

**Tuple packing** means placing multiple values together in one tuple. Python automatically creates a tuple when several comma-separated values are assigned to a single variable. Parentheses can be used to make the tuple easier to read.

### Syntax

```python
tuple_name = value1, value2, value3
```

Example:


````python
# Packing three values into one tuple
student_details = "Maya", 20, "BSc IT"

print(student_details)
print(type(student_details))
````

Here, the three values are packed into `student_details`. Even though parentheses are not used, Python recognizes the comma-separated values as a tuple.

## 3. Tuple unpacking

**Tuple unpacking** is the opposite of packing. It means assigning the values stored in a tuple to separate variables. The number of variables should normally match the number of items in the tuple.

### Syntax

```python
variable1, variable2, variable3 = tuple_name
```

Example:


````python
student_details = ("Maya", 20, "BSc IT")

# Unpacking the tuple
name, age, course = student_details

print("Name:", name)
print("Age:", age)
print("Course:", course)
````

In this example, each value is assigned to its matching variable. This makes it easier to work with individual items without accessing them by their index.

Python also supports using `*` to collect extra values into a list during unpacking.

````python
numbers = (10, 20, 30, 40, 50)

first, *middle, last = numbers

print("First value:", first)
print("Middle values:", middle)
print("Last value:", last)
````

## 4. Tuple methods

Tuples cannot be modified after creation, so they have only a small number of built-in methods. Two commonly used methods are `count()` and `index()`.

### 4.1 `count()` method

The `count()` method returns how many times a particular value appears in a tuple.

**Syntax:**

```python
tuple_name.count(value)
```

Example:


````python
marks = (75, 80, 90, 75, 85, 75)

number_of_75s = marks.count(75)

print("The value 75 appears", number_of_75s, "times.")
````

The value `75` appears three times in the tuple, so `count(75)` returns `3`. This method is useful when we need to find the frequency of a value.

### 4.2 `index()` method

The `index()` method returns the position of the first occurrence of a specified value in a tuple. Python indexing starts from `0`, not `1`.

**Syntax:**

```python
tuple_name.index(value)
```

Example:


````python
subjects = ("Python", "Networking", "Database", "Mathematics")

position = subjects.index("Database")

print("Database is at index:", position)
````

The output is `2` because `"Database"` is the third item, and its index is 2. If the requested value is not present in the tuple, `index()` raises a `ValueError`.

### Quick comparison of tuple methods

| Method | Purpose |
|---|---|
| `count(value)` | Counts how many times a value appears. |
| `index(value)` | Returns the index of the first occurrence of a value. |

## 5. Sets

A **set** is a built-in Python data type used to store multiple unique values in one variable. Sets are unordered, which means their items do not have fixed positions and cannot be accessed using indexes. A set automatically removes duplicate values.

Sets are useful when we need to work with unique items, such as a list of registered student IDs or the different subjects selected by students.

### Basic syntax

```python
set_name = {item1, item2, item3}
```

Example:


````python
subjects = {"Python", "Networking", "Database", "Python"}

print(subjects)
````

Although `"Python"` was entered twice, it appears only once in the set. The order of the displayed items may be different because sets are unordered.

**Note:** To create an empty set, use `set()`. Writing `{}` creates an empty dictionary, not an empty set.

````python
empty_set = set()
print(empty_set)
print(type(empty_set))
````

## 6. Set methods

Unlike tuples, sets are mutable. This means we can add new items and remove existing items after creating a set. Python provides several methods to perform these operations.

### 6.1 `add()` method

The `add()` method inserts one new item into a set. If the item is already present, the set remains unchanged because sets do not allow duplicates.

**Syntax:**

```python
set_name.add(value)
```

Example:


````python
fruits = {"Apple", "Banana", "Mango"}

fruits.add("Orange")

print(fruits)
````

The `add()` method inserts `"Orange"` into the set. The method changes the original set directly and does not return a new set.

### 6.2 `update()` method

The `update()` method adds multiple items from another iterable, such as a list, tuple, or another set. It is useful when we want to add several values at once.

**Syntax:**

```python
set_name.update(iterable)
```

Example:


````python
fruits = {"Apple", "Banana"}

new_fruits = ["Mango", "Orange", "Grapes"]

fruits.update(new_fruits)

print(fruits)
````

Here, all the items from `new_fruits` are added to the `fruits` set. Any duplicate values are ignored automatically.

### 6.3 `discard()` method

The `discard()` method removes a specified item from a set. If the item does not exist, it does not produce an error.

**Syntax:**

```python
set_name.discard(value)
```

Example:


````python
colors = {"Red", "Blue", "Green", "Yellow"}

colors.discard("Blue")
colors.discard("Black")  # This value is not in the set

print(colors)
````

In this example, `"Blue"` is removed from the set. `"Black"` is not present, but using `discard()` does not cause an error. This makes it useful when we are not sure whether an item exists.

### 6.4 `remove()` method

The `remove()` method also removes a specified item from a set. However, unlike `discard()`, it raises a `KeyError` if the item is not found.

**Syntax:**

```python
set_name.remove(value)
```

Example:


````python
colors = {"Red", "Blue", "Green", "Yellow"}

colors.remove("Green")

print(colors)
````

The value `"Green"` is removed from the set. We should use `remove()` when we expect the item to be present. If the item might not exist, `discard()` is usually safer.

### Difference between `discard()` and `remove()`

| Method | If the item exists | If the item does not exist |
|---|---|---|
| `discard()` | Removes the item | No error |
| `remove()` | Removes the item | Raises `KeyError` |

### 6.5 `union()` method

The `union()` method combines the items from two or more sets into a new set. Duplicate items are included only once. The original sets are not changed.

**Syntax:**

```python
set1.union(set2)
```

The `|` operator can also be used to find the union of sets.

Example:


````python
python_students = {"Aarav", "Maya", "Rohan"}
java_students = {"Maya", "Sita", "Rohan"}

all_students = python_students.union(java_students)

print("Students in either class:", all_students)
````

The union contains every unique student from both sets. `"Maya"` and `"Rohan"` appear in both original sets, but each appears only once in the result.

### 6.6 `intersection()` method

The `intersection()` method returns a new set containing only the items that are common to both sets. It is useful when we want to find shared values.

**Syntax:**

```python
set1.intersection(set2)
```

The `&` operator can also be used to find the intersection of sets.

Example:


````python
python_students = {"Aarav", "Maya", "Rohan"}
java_students = {"Maya", "Sita", "Rohan"}

common_students = python_students.intersection(java_students)

print("Students in both classes:", common_students)
````

The result contains `"Maya"` and `"Rohan"` because they are present in both sets. The original sets remain unchanged.

### Quick comparison of set methods

| Method | Purpose |
|---|---|
| `add(value)` | Adds one item to a set. |
| `update(iterable)` | Adds multiple items from an iterable. |
| `discard(value)` | Removes an item without an error if it is missing. |
| `remove(value)` | Removes an item but raises `KeyError` if it is missing. |
| `union(other_set)` | Returns all unique items from both sets. |
| `intersection(other_set)` | Returns only the items common to both sets. |

## 7. Tuples vs. sets

Although tuples and sets can both store multiple values, they are used for different purposes.

| Feature | Tuple | Set |
|---|---|---|
| Order | Ordered | Unordered |
| Duplicate values | Allowed | Not allowed |
| Change after creation | Immutable | Mutable |
| Indexing | Supported | Not supported |
| Main use | Keeping a fixed collection of values | Keeping unique values and comparing collections |
| Example | `("Aarav", 20, "IT")` | `{"Python", "Java", "C"}` |

## 8. Practical example

The following example uses both tuples and sets in a simple student-record situation. A tuple stores a student's fixed details, while sets store the students enrolled in two different courses. The union and intersection methods help compare the course groups.


````python
student = ("Maya", 20, "BSc IT")

name, age, course = student

print("Student Information")
print("Name:", name)
print("Age:", age)
print("Course:", course)

python_class = {"Maya", "Aarav", "Rohan"}
networking_class = {"Maya", "Sita", "Rohan"}

all_students = python_class.union(networking_class)
students_in_both = python_class.intersection(networking_class)

print("\nAll enrolled students:", all_students)
print("Students in both classes:", students_in_both)
````

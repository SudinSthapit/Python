## 1. Introduction to dictionaries

A **dictionary** is a Python data structure that stores information as key-value pairs. It is written using curly brackets `{}`, with a colon (`:`) separating each key from its value and commas separating the pairs.

Dictionaries are **mutable**, which means their contents can be changed after they are created. Keys must be unique, while values can be repeated. Dictionaries preserve insertion order in modern Python versions.

### Basic syntax

```python
dictionary_name = {
    "key1": "value1",
    "key2": "value2"
}
```

### Example


````python
student = {
    "name": "Aarav",
    "age": 19,
    "course": "Information Technology"
}

print(student)
print(type(student))
````

In this example, `name`, `age`, and `course` are the keys. Their corresponding values are `"Aarav"`, `19`, and `"Information Technology"`. The `type()` function confirms that `student` is a dictionary.

## 2. Key-value pairs

A key-value pair has two parts:

- **Key:** The label used to identify an item.
- **Value:** The information stored under that key.

For example, in `"name": "Aarav"`, the key is `"name"` and the value is `"Aarav"`.

A dictionary can store different types of values, such as strings, numbers, lists, tuples, Boolean values, and even other dictionaries. Keys should be immutable types, such as strings, numbers, or tuples containing immutable values.

### Example with different value types


````python
student = {
    "name": "Maya",
    "age": 20,
    "marks": 85.5,
    "subjects": ["Python", "Networking", "Database"],
    "is_enrolled": True
}

print(student)
````

This dictionary stores a string, an integer, a decimal number, a list, and a Boolean value. This shows that dictionary values do not all need to have the same data type.

## 3. Accessing dictionary values

We can access a value by writing its key inside square brackets `[]`. Unlike a list, we do not need to remember an item's numerical position. We just need to know its key.

### Syntax

```python
dictionary_name["key"]
```

### Example


````python
student = {
    "name": "Maya",
    "age": 20,
    "course": "BSc IT"
}

print("Student name:", student["name"])
print("Student age:", student["age"])
print("Student course:", student["course"])
````

The expression `student["name"]` returns `"Maya"`. If we try to access a key that does not exist using square brackets, Python raises a `KeyError`. The `get()` method is useful when we want to avoid this error.

## 4. Adding and updating dictionary values

Dictionaries are mutable, so we can add new key-value pairs or change existing values after creating them.

### Adding a new item

Assign a value to a key that is not already in the dictionary.

**Syntax:**

```python
dictionary_name["new_key"] = value
```

Example:


````python
student = {
    "name": "Aarav",
    "age": 19
}

student["course"] = "Information Technology"

print(student)
````

The key `"course"` was not present before, so Python adds it with the value `"Information Technology"`.

### Updating an existing item

Assign a new value to a key that already exists.

**Syntax:**

```python
dictionary_name["existing_key"] = new_value
```

Example:


````python
student = {
    "name": "Aarav",
    "age": 19,
    "course": "BSc CSIT"
}

student["course"] = "BSc IT"

print(student)
````

The course value changes from `"BSc CSIT"` to `"BSc IT"`. The other items remain unchanged.

## 5. Nested dictionaries

A **nested dictionary** is a dictionary stored inside another dictionary. It is useful for grouping related information. For example, a main dictionary can store details of several students, with each student's information kept in a separate inner dictionary.

### Basic syntax

```python
main_dictionary = {
    "outer_key": {
        "inner_key": "value"
    }
}
```

### Example


````python
students = {
    "student1": {
        "name": "Aarav",
        "age": 19,
        "course": "BSc IT"
    },
    "student2": {
        "name": "Maya",
        "age": 20,
        "course": "BCA"
    }
}

print(students)
````

Here, `students` is the main dictionary. It contains two inner dictionaries, `student1` and `student2`, and each one stores a student's details.

### Accessing values in a nested dictionary

Use the outer key first, followed by the inner key.

**Syntax:**

```python
dictionary_name["outer_key"]["inner_key"]
```

Example:


````python
print("First student's name:", students["student1"]["name"])
print("Second student's course:", students["student2"]["course"])
````

We first select a student using the outer key and then access the required detail using the inner key.

### Updating a nested dictionary

We can update or add information inside an inner dictionary by referring to both keys.

Example:


````python
students["student1"]["age"] = 20
students["student2"]["email"] = "maya@example.com"

print(students)
````

The first student's age is updated, and an email address is added to the second student's details.

## 6. Dictionary methods

Python provides built-in methods to view, access, update, and remove dictionary information. The following are the methods required for this topic.

### 6.1 `keys()` method

The `keys()` method returns a view containing all the keys in a dictionary.

**Syntax:**

```python
dictionary_name.keys()
```

Example:


````python
student = {
    "name": "Maya",
    "age": 20,
    "course": "BSc IT"
}

print(student.keys())
print(list(student.keys()))
````

The method displays all the keys: `name`, `age`, and `course`. It returns a dictionary view, which can be converted into a list using `list()`.

### 6.2 `values()` method

The `values()` method returns a view containing all the values stored in a dictionary.

**Syntax:**

```python
dictionary_name.values()
```

Example:


````python
print(student.values())
print(list(student.values()))
````

This displays the values stored in the dictionary. It is useful when we need the information but do not need the keys.

### 6.3 `items()` method

The `items()` method returns a view containing all key-value pairs. Each pair is returned as a tuple.

**Syntax:**

```python
dictionary_name.items()
```

Example:


````python
print(student.items())

for key, value in student.items():
    print(key, ":", value)
````

The `for` loop uses `items()` to get each key and value together, making it easy to display the dictionary in a readable way.

### 6.4 `get()` method

The `get()` method accesses a value using its key. If the key does not exist, it returns `None` by default instead of raising a `KeyError`. We can also provide a default value.

**Syntax:**

```python
dictionary_name.get(key)
dictionary_name.get(key, default_value)
```

Example:


````python
student = {
    "name": "Aarav",
    "age": 19,
    "course": "Information Technology"
}

print(student.get("name"))
print(student.get("marks"))
print(student.get("marks", "Marks not available"))
````

The first statement returns `"Aarav"`. The second returns `None` because `marks` is not a key in the dictionary. The third returns the default message `"Marks not available"`.

### 6.5 `update()` method

The `update()` method adds new key-value pairs or changes existing values. It can take another dictionary or an iterable of key-value pairs.

**Syntax:**

```python
dictionary_name.update(other_dictionary)
```

Example:


````python
student = {
    "name": "Maya",
    "age": 20
}

new_details = {
    "course": "BSc IT",
    "age": 21
}

student.update(new_details)

print(student)
````

The `course` key is added, and the existing `age` value is changed from `20` to `21`. The `update()` method changes the original dictionary directly.

### 6.6 `pop()` method

The `pop()` method removes an item using its key and returns the value that was removed. If the key is missing, it raises a `KeyError` unless a default value is provided.

**Syntax:**

```python
dictionary_name.pop(key)
dictionary_name.pop(key, default_value)
```

Example:


````python
student = {
    "name": "Aarav",
    "age": 19,
    "course": "BSc IT"
}

removed_value = student.pop("age")

print("Removed age:", removed_value)
print("Updated dictionary:", student)
````

The `pop("age")` statement removes the `age` key and returns its value, `19`. The remaining items stay in the dictionary.

We can also provide a default value so that a missing key does not cause an error.

````python
student = {
    "name": "Aarav",
    "course": "BSc IT"
}

result = student.pop("age", "Age not found")

print(result)
print(student)
````

Since `age` is not present, `pop()` returns the default message instead of raising an error.

## 7. Summary of dictionary methods

| Method | Purpose |
|---|---|
| `keys()` | Returns a view of all dictionary keys. |
| `values()` | Returns a view of all dictionary values. |
| `items()` | Returns a view of all key-value pairs as tuples. |
| `get(key)` | Returns the value for a key, or `None` if it is missing. |
| `update()` | Adds new key-value pairs or updates existing ones. |
| `pop(key)` | Removes a key and returns its associated value. |

## 8. Practical example: Student information system

The following example combines the concepts covered in this notebook. It stores student details, accesses information, updates values, displays the keys and values, and removes an item using `pop()`.


````python
student = {
    "name": "Maya",
    "age": 20,
    "course": "BSc IT",
    "marks": 85
}

# Accessing information with get()
print("Student name:", student.get("name"))
print("Student marks:", student.get("marks"))

# Adding and updating information
student.update({
    "age": 21,
    "semester": 2
})

# Displaying all key-value pairs
print("\nUpdated student details:")
for key, value in student.items():
    print(key, ":", value)

# Displaying keys and values
print("\nKeys:", list(student.keys()))
print("Values:", list(student.values()))

# Removing one item
removed_marks = student.pop("marks")

print("\nRemoved marks:", removed_marks)
print("Final student details:", student)
````

This example shows how dictionaries can be used to manage student records. The `get()` method accesses information safely, `update()` adds or changes details, `items()` displays each key with its value, and `pop()` removes a selected item.

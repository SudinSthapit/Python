# Modules and Packages in Python

**Topic:** `import`, `from ... import`, aliases, custom modules, packages, `__init__.py`, `math`, `random`, `datetime`, and `pip`

## Introduction

As Python programs get larger, it is useful to divide code into smaller files and group related files together. Python provides **modules** and **packages** to help organize code and reuse work instead of writing the same functions again and again.

A module is usually a Python file, while a package is a directory that groups related modules. Python also includes useful standard-library modules, such as `math`, `random`, and `datetime`. The `pip` tool helps install third-party packages.

This notebook explains the required concepts with simple examples and short notes. Run the code cells in order to see the results.


## 1. What Is a Module?

A **module** is a Python file that contains code such as variables, functions, and classes. Its filename usually ends with `.py`.

For example, a file named `calculator.py` can contain functions for addition and subtraction. Another Python file can import that module and use its functions.

### Example syntax

```python
import module_name
```

After importing a module, use a dot (`.`) to access its contents, such as `module_name.function_name()`.


````python
import math

print("Square root of 81:", math.sqrt(81))
print("Value of pi:", math.pi)
print("Rounded up:", math.ceil(4.2))
````

In this example, `math` is a standard-library module. The functions and values are accessed with the `math.` prefix, which makes it clear where they come from.


## 2. Using `import`

The `import` statement loads a module so that its functions, classes, and variables can be used in the current program.

### Syntax

```python
import module_name
import module_one, module_two
```

It is often clearer to put separate imports on separate lines.


````python
import math
import random

number = 25
print("Square root:", math.sqrt(number))
print("Random integer from 1 to 10:", random.randint(1, 10))
````

`math.sqrt()` calculates a square root, while `random.randint(1, 10)` returns a random integer from 1 through 10, including both endpoints.


## 3. Using `from ... import`

The `from ... import` form imports selected names from a module. This lets you use those names directly without writing the module prefix each time.

### Syntax

```python
from module_name import name
from module_name import name_one, name_two
```

Importing only the names you need can make a short example convenient. However, using the module prefix can make the origin of a function easier to recognize.


````python
from math import sqrt, pi

print("Square root of 49:", sqrt(49))
print("Half of pi:", pi / 2)
````

Because `sqrt` and `pi` were imported directly, they can be used by name. We do not need to write `math.sqrt()` or `math.pi` in this cell.


## 4. Importing with Aliases

An **alias** is an alternative name given to a module or imported item. The `as` keyword creates the alias.

### Syntax

```python
import module_name as short_name
from module_name import function_name as short_name
```

Aliases can make a commonly used module name shorter. Python programmers often use `import numpy as np` when working with NumPy.


````python
import datetime as dt
from math import factorial as fact

today = dt.date.today()
print("Today's date:", today)
print("5 factorial:", fact(5))
````

Here, `dt` is used instead of `datetime`, and `fact` is used instead of `factorial`. The aliases only change the names used in this file; they do not rename the original module or function.


## 5. Creating a Custom Module

A **custom module** is a Python file that you create yourself. It can hold functions, variables, or classes that you want to reuse in other files.

Suppose a folder contains these two files:

**`my_tools.py`**
```python
def greet(name):
    return f"Hello, {name}!"

def multiply(a, b):
    return a * b

course = "Python"
```

**`main.py`**
```python
import my_tools

print(my_tools.greet("Sita"))
print(my_tools.multiply(4, 5))
print(my_tools.course)
```

Save both files in the same folder for this simple example. Then run `main.py`. Python can find `my_tools.py` because it is in the same directory as the script being run (assuming no other path or import issue).


### A small demonstration in this notebook

A notebook cell is not automatically a separate `.py` module. The example below demonstrates the same function idea directly; to practise importing a custom module, create the two files shown above in a folder and run them there.


````python
def greet(name):
    return f"Hello, {name}!"

def multiply(a, b):
    return a * b

print(greet("Sita"))
print("4 multiplied by 5 =", multiply(4, 5))
````

## 6. Packages

A **package** is a directory that groups related Python modules. Packages help keep larger projects organized.

For example, a project could have this structure:

```text
school_project/
│
├── main.py
└── student_tools/
    ├── __init__.py
    ├── marks.py
    └── attendance.py
```

The `student_tools` directory is the package. The `marks.py` and `attendance.py` files are modules inside it.

### Import syntax

```python
import student_tools.marks
from student_tools import marks
from student_tools.marks import calculate_average
```

Python also supports namespace packages that do not contain an `__init__.py` file. For a beginner's regular package, including `__init__.py` is a common and useful practice.


## 7. The Purpose of `__init__.py`

The `__init__.py` file is placed inside a package directory. In a regular package, it marks the directory as a package and can contain package initialization code or expose selected names for convenient imports.

Example project:

```text
student_tools/
├── __init__.py
└── marks.py
```

**`marks.py`**
```python
def calculate_average(marks):
    return sum(marks) / len(marks)
```

**`__init__.py`**
```python
from .marks import calculate_average
```

With this setup, code can use:

```python
from student_tools import calculate_average

print(calculate_average([80, 75, 90]))
```

The leading dot in `from .marks import ...` means the `marks` module is being imported relative to the current package.


## 8. The `math` Module

The `math` module provides common mathematical constants and functions. It is included with Python, so it does not need to be installed separately.

### Common constants and functions

| Name | Purpose | Example |
|---|---|---|
| `math.pi` | Approximation of pi | `math.pi` |
| `math.sqrt(x)` | Square root of `x` | `math.sqrt(64)` |
| `math.ceil(x)` | Rounds up to an integer | `math.ceil(3.2)` |
| `math.floor(x)` | Rounds down to an integer | `math.floor(3.8)` |
| `math.factorial(n)` | Factorial of a non-negative integer | `math.factorial(5)` |
| `math.pow(x, y)` | Raises `x` to power `y` and returns a float | `math.pow(2, 3)` |
| `math.fabs(x)` | Absolute value as a float | `math.fabs(-7)` |
| `math.sin(x)` | Sine of an angle in radians | `math.sin(math.pi / 2)` |


````python
import math

print("Pi:", math.pi)
print("Square root of 64:", math.sqrt(64))
print("Ceiling of 3.2:", math.ceil(3.2))
print("Floor of 3.8:", math.floor(3.8))
print("Factorial of 5:", math.factorial(5))
print("2 raised to 3:", math.pow(2, 3))
print("Absolute value of -7:", math.fabs(-7))
print("Sine of pi/2:", math.sin(math.pi / 2))
````

**Note:** Trigonometric functions in `math` expect angles in radians, not degrees. For example, `math.pi / 2` radians equals 90 degrees.


## 9. The `random` Module

The `random` module provides functions for generating pseudo-random values. It is useful for simple games, simulations, and choosing random items. It is not suitable for passwords, security tokens, or other security-sensitive tasks; use the `secrets` module for those cases.

### Common functions

| Function | Purpose |
|---|---|
| `random.random()` | Returns a float from 0.0 up to, but not including, 1.0 |
| `random.randint(a, b)` | Returns an integer from `a` through `b`, inclusive |
| `random.randrange(start, stop)` | Selects from the integer range, excluding `stop` |
| `random.choice(sequence)` | Picks one item from a non-empty sequence |
| `random.shuffle(list)` | Shuffles a list in place |
| `random.uniform(a, b)` | Returns a random floating-point number between the bounds |

The output may change each time the cell is run.


````python
import random

print("Random decimal:", random.random())
print("Random number from 1 to 6:", random.randint(1, 6))
print("Random even number from 2 to 10:", random.randrange(2, 11, 2))

colours = ["red", "blue", "green", "yellow"]
print("Random colour:", random.choice(colours))

cards = ["A", "K", "Q", "J"]
random.shuffle(cards)
print("Shuffled cards:", cards)

print("Random decimal from 1 to 5:", random.uniform(1, 5))
````

If you need repeatable results while learning or testing, you can set a seed with `random.seed(10)`. The same seed can reproduce a sequence in the same Python environment, but it should not be used as a security feature.


## 10. The `datetime` Module

The `datetime` module provides classes for working with dates and times. It is useful for displaying today's date, recording times, comparing dates, and calculating time differences.

### Common classes and methods

| Name | Purpose |
|---|---|
| `date.today()` | Gets the current local date |
| `datetime.now()` | Gets the current local date and time |
| `date(year, month, day)` | Creates a date |
| `timedelta(days=...)` | Represents a duration or difference |
| `strftime(format)` | Formats a date/time as text |
| `strptime(text, format)` | Parses text into a date/time object |

Example format codes include `%Y` for a four-digit year, `%m` for month, `%d` for day, `%H` for hour, and `%M` for minute.


````python
from datetime import date, datetime, timedelta

today = date.today()
current_time = datetime.now()

print("Today's date:", today)
print("Current local date and time:", current_time)

birthday = date(2005, 7, 15)
print("Example birthday:", birthday)

next_week = today + timedelta(days=7)
print("Date seven days from now:", next_week)

print("Formatted date:", today.strftime("%d-%m-%Y"))
print("Formatted time:", current_time.strftime("%H:%M"))
````

The current date and time depend on the computer or environment running the notebook. `strftime()` converts a date or time to a formatted string, while `strptime()` does the reverse by reading a string that matches a format.


````python
from datetime import datetime

date_text = "02-11-2026"
parsed_date = datetime.strptime(date_text, "%d-%m-%Y")

print("Original text:", date_text)
print("Parsed date:", parsed_date)
print("Year:", parsed_date.year)
print("Month:", parsed_date.month)
print("Day:", parsed_date.day)
````

## 11. What Is `pip`?

`pip` is the standard package installer commonly used to install and manage third-party Python packages from the Python Package Index (PyPI) and other package sources. Packages such as `requests` or `pandas` are not part of Python's core standard library, so they may need to be installed separately.

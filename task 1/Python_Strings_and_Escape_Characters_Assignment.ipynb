## Introduction

Strings are one of the most commonly used data types in Python. They are used to store text, such as names, messages, addresses, and other information. Python provides different ways to access and format string characters, as well as built-in methods that make it easier to work with text. In this assignment, I will explain how strings work, how indexing and slicing are used, how f-strings and escape characters help format text, and how common string methods can be used in practical examples.

## 1. Introduction to Strings

A string is a sequence of characters enclosed in single quotation marks, double quotation marks, or triple quotation marks. It can contain letters, numbers, spaces, and special characters. Strings are useful whenever a program needs to store or display text.

**Basic syntax:**

```python
variable_name = "some text"
```

For example, a student's name or a course title can be stored in a string variable.

````python
student_name = "Alex"
course = 'Information Technology'
message = "Welcome to Python programming!"

print(student_name)
print(course)
print(message)
print(type(student_name))
````

**Explanation:** Each variable stores text, so Python identifies its data type as `str`. Strings can be written using either single or double quotation marks, as long as the opening and closing quotation marks match.

## 2. String Indexing

Indexing is used to access an individual character from a string. Python starts counting positions from `0`, so the first character has index `0`, the second has index `1`, and so on. Python also supports negative indexing, which starts counting from the end of the string. The index `-1` refers to the last character.

**Basic syntax:**

```python
string_name[index]
```

For example, if `word = "Python"`, then `word[0]` returns `P` and `word[-1]` returns `n`.

````python
word = "Python"

print("First character:", word[0])
print("Second character:", word[1])
print("Last character:", word[-1])
print("Second last character:", word[-2])
````

**Explanation:** The output shows how positive and negative indexes access characters in the same string. If we try to access an index that does not exist, Python raises an `IndexError`. For example, `word[10]` is invalid because the word has only six characters.

## 3. String Slicing

Slicing is used to take a part of a string instead of just one character. It allows us to select a range of characters by using a starting index and an ending index. The starting index is included, but the ending index is not included. We can also provide a step value to skip characters.

**Basic syntax:**

```python
string_name[start:end:step]
```

The start, end, and step values are optional. If the start is omitted, Python starts from the beginning. If the end is omitted, Python continues to the end of the string.

````python
language = "Programming"

print("Characters from index 0 to 5:", language[0:6])
print("From index 3 to the end:", language[3:])
print("From the beginning to index 4:", language[:5])
print("Every second character:", language[::2])
print("Reversed string:", language[::-1])
````

**Explanation:** The slice `language[0:6]` returns the characters from index 0 up to, but not including, index 6. The expression `language[::2]` selects every second character. The expression `language[::-1]` uses a negative step to return the string in reverse order. Slicing is useful when we need to extract part of a name, word, or other text.

## 4. f-Strings (Formatted String Literals)

An f-string is a simple way to place variable values or expressions inside a string. It starts with the letter `f` before the quotation mark. Inside the string, we use curly brackets `{}` around the variable or expression that we want Python to evaluate. F-strings make output easier to read and are commonly used when displaying information.

**Basic syntax:**

```python
f"Some text {variable_name}"
```

````python
name = "Sam"
age = 20
course = "Python Programming"

print(f"My name is {name}.")
print(f"I am {age} years old.")
print(f"I am studying {course}.")
print(f"Next year, I will be {age + 1} years old.")
````

**Explanation:** The values inside curly brackets are replaced with the actual variable values. We can also place simple calculations, such as `age + 1`, inside the brackets. This is often clearer than joining text and variables with the `+` operator.

## 5. Escape Characters

Escape characters are special character combinations that begin with a backslash (`\\`). They are used to include special formatting or characters inside a string. For example, they can create a new line, add a tab space, or allow quotation marks to appear inside a string.

Some commonly used escape characters are:

| Escape character | Meaning | Example |
|---|---|---|
| `\\n` | New line | Moves text to the next line |
| `\\t` | Horizontal tab | Adds a tab space |
| `\\'` | Single quotation mark | Displays `'` inside a string |
| `\\"` | Double quotation mark | Displays `"` inside a string |
| `\\\\` | Backslash | Displays one backslash character |

Escape characters are useful when formatting output or displaying text that contains quotation marks and other special characters.

### 5.1 New Line (`\\n`)

The `\\n` escape character moves the text that follows it to a new line. It is useful when we want to display information on separate lines without using multiple `print()` statements.

````python
message = "Hello!\\nWelcome to Python.\\nLet's start learning."
print(message)
````

**Explanation:** Each `\\n` creates a line break. As a result, the message appears on three separate lines when printed.

### 5.2 Tab (`\\t`)

The `\\t` escape character adds a horizontal tab space. It is useful for separating pieces of information and making simple output easier to read.

````python
print("Name:\\tAlex")
print("Course:\\tInformation Technology")
print("Level:\\tFirst Year")
````

**Explanation:** The `\\t` character creates a tab space between each label and its value. The exact spacing can depend on the notebook or terminal where the code is displayed.

### 5.3 Quotation Marks Inside Strings

Sometimes we need to include quotation marks inside a string. We can use a different type of quotation mark around the string, or place a backslash before the quotation mark that we want to display. This prevents Python from treating that quotation mark as the end of the string.

````python
single_quote = "It's a good day."
double_quote = 'He said, "Hello!"'
escaped_quote = "She said, \"Python is easy!\""

print(single_quote)
print(double_quote)
print(escaped_quote)
````

**Explanation:** The first two examples use different quotation marks to avoid confusion. In the third example, a backslash is placed before the double quotation marks so they can be displayed as part of the string.

### 5.4 Backslash (`\\\\`)

A backslash is used to introduce escape characters. If we want to display an actual backslash in a normal Python string, we write two backslashes (`\\\\`). Python interprets the pair as one visible backslash.

````python
folder_path = "C:\\\\Users\\\\Student\\\\Documents"
print(folder_path)
````

**Explanation:** Each pair of backslashes in the code produces one backslash in the output. This is useful when writing Windows-style file paths in normal Python strings. Another option for many file paths is to use a raw string, such as `r"C:\\Users\\Student"`.

## 6. Common String Methods

Python provides built-in string methods that make it easier to work with text. A method is called by writing the string variable, a dot, the method name, and parentheses. Most string methods return a new string instead of changing the original string because strings in Python are immutable. This means that the original string cannot be changed directly.

The methods covered in this assignment are `strip()`, `split()`, `join()`, `replace()`, `find()`, `upper()`, and `lower()`.

### 6.1 `strip()` Method

The `strip()` method removes whitespace from the beginning and end of a string. Whitespace can include spaces, tabs, and new line characters. It does not remove spaces from between words.

**Syntax:**

```python
string_name.strip()
```

````python
user_input = "   Hello Python!   "
clean_text = user_input.strip()

print("Original text:", repr(user_input))
print("After strip():", repr(clean_text))
````

**Explanation:** The original text contains extra spaces at the beginning and end. The `strip()` method removes those outer spaces, while keeping the space between `Hello` and `Python`. The `repr()` function is used here to make the remaining spaces easier to see in the output.

### 6.2 `split()` Method

The `split()` method divides a string into a list of smaller strings. By default, it separates the text wherever there is whitespace. We can also provide a separator, such as a comma, to control where the string is divided.

**Syntax:**

```python
string_name.split(separator)
```

The separator is optional. If it is not provided, Python splits the string using whitespace.

````python
sentence = "Python is easy to learn"
words = sentence.split()

subjects_text = "Python,Networking,Database"
subjects = subjects_text.split(",")

print("Words:", words)
print("Subjects:", subjects)
````

**Explanation:** The first `split()` separates the sentence at whitespace and returns a list of words. The second uses a comma as the separator, so each subject becomes a separate list item. This method is useful when processing user input or information stored as text.

### 6.3 `join()` Method

The `join()` method combines the items in an iterable, such as a list of strings, into one string. The string before `.join()` is used as the separator between the items.

**Syntax:**

```python
separator.join(iterable)
```

````python
words = ["Python", "is", "useful"]

sentence = " ".join(words)
comma_separated = ", ".join(words)

print("Joined sentence:", sentence)
print("Comma separated:", comma_separated)
````

**Explanation:** The space in `" ".join(words)` is placed between each word. In the second example, a comma and a space are used as the separator. The items being joined should be strings; non-string items need to be converted first, for example with `str()`.

### 6.4 `replace()` Method

The `replace()` method creates a new string in which one piece of text is replaced with another. It is useful for correcting words, changing labels, or updating text. The original string remains unchanged.

**Syntax:**

```python
string_name.replace(old_value, new_value)
```

An optional third argument can limit the number of replacements.

````python
message = "I like Java. Java is popular."
updated_message = message.replace("Java", "Python")

print("Original:", message)
print("Updated:", updated_message)
````

**Explanation:** The `replace()` method changes each occurrence of `Java` to `Python` in the new string. The original variable `message` still contains the original text because strings are immutable.

### 6.5 `find()` Method

The `find()` method searches for a word or character inside a string. It returns the index of the first occurrence when the text is found. If the text is not found, it returns `-1` instead of raising an error.

**Syntax:**

```python
string_name.find(substring)
```

````python
sentence = "Python programming is interesting"

position = sentence.find("programming")
missing_position = sentence.find("Java")

print("Position of programming:", position)
print("Position of Java:", missing_position)
````

**Explanation:** The word `programming` is found in the sentence, so `find()` returns the index where it starts. The word `Java` is not present, so the method returns `-1`. The search is case-sensitive, meaning uppercase and lowercase letters must match.

### 6.6 `upper()` Method

The `upper()` method converts all lowercase letters in a string into uppercase letters. It is useful when we want text to appear in capital letters, such as headings or labels.

**Syntax:**

```python
string_name.upper()
```

````python
course = "information technology"
capital_course = course.upper()

print("Original:", course)
print("Uppercase:", capital_course)
````

**Explanation:** The `upper()` method changes the letters in the returned string to uppercase. The original variable `course` remains unchanged.

### 6.7 `lower()` Method

The `lower()` method converts uppercase letters in a string into lowercase letters. It is useful when comparing text without being affected by differences in capitalization.

**Syntax:**

```python
string_name.lower()
```

````python
greeting = "HELLO PYTHON"
small_greeting = greeting.lower()

print("Original:", greeting)
print("Lowercase:", small_greeting)
````

**Explanation:** The `lower()` method converts the uppercase letters into lowercase letters. It is often helpful when checking user input because it can make text comparisons more consistent. For example, `input_text.lower() == "yes"` can accept `yes`, `YES`, or `Yes`.

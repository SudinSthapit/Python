# Python Lists: Creating, Indexing, Slicing, and List Methods

topic: Lists in Python  


## 1. Introduction

A list is a built-in Python data type used to store more than one item in a single variable. Lists are useful when we want to keep related values together, such as student names, marks, or items in a shopping list.

A list is ordered, changeable (mutable), and can contain duplicate values. It can also hold values of different data types, although it is often clearer to keep similar kinds of data in one list.

In this notebook, I will demonstrate how to create lists, access and update their items, and use common list methods.

## 2. Creating a List

Lists are written using square brackets `[]`. Items are separated by commas.

**Syntax**
```python
list_name = [item1, item2, item3]
```

A list can contain strings, numbers, or a mixture of data types.

````python
# Creating a few example lists
fruits = ["apple", "banana", "mango", "orange"]
marks = [78, 85, 91, 67]
mixed_items = ["Sam", 19, 3.5, True]

print("Fruits:", fruits)
print("Marks:", marks)
print("Mixed items:", mixed_items)
````

## 3. Indexing a List

Indexing is used to access a single item by its position. Python indexing starts at `0`, so the first item is at index `0`. Negative indexing counts from the end: `-1` means the last item.

**Syntax**
```python
list_name[index]
```

````python
fruits = ["apple", "banana", "mango", "orange"]

print(fruits[0])   # First item
print(fruits[2])   # Third item
print(fruits[-1])  # Last item
````

## 4. Slicing a List

Slicing is used to get a section of a list. The ending index is not included in the result.

**Syntax**
```python
list_name[start:stop:step]
```

- `start` is the index where the slice begins.
- `stop` is the index where the slice ends, but that item is not included.
- `step` tells Python how many positions to move each time. It is optional.

````python
fruits = ["apple", "banana", "mango", "orange", "grapes"]

print(fruits[1:4])  # Items at indexes 1, 2, and 3
print(fruits[:3])   # From the beginning up to index 2
print(fruits[2:])   # From index 2 to the end
print(fruits[::2])  # Every second item
````

## 5. Updating List Items

Lists are mutable, which means that their items can be changed after the list has been created. To update an item, assign a new value to its index.

**Syntax**
```python
list_name[index] = new_value
```

````python
subjects = ["English", "Math", "Science"]
subjects[1] = "Computer"

print(subjects)
````

## 6. Adding Items to a List

### `append()`

The `append()` method adds one item to the end of a list.

**Syntax**
```python
list_name.append(item)
```

````python
fruits = ["apple", "banana"]
fruits.append("mango")

print(fruits)
````

### `extend()`

The `extend()` method adds each item from another iterable (such as another list) to the end of the current list.

**Syntax**
```python
list_name.extend(iterable)
```

````python
fruits = ["apple", "banana"]
more_fruits = ["mango", "orange"]

fruits.extend(more_fruits)
print(fruits)
````

### `insert()`

The `insert()` method adds an item at a chosen index. The existing items at that index and after it move one position to the right.

**Syntax**
```python
list_name.insert(index, item)
```

````python
colours = ["red", "blue", "green"]
colours.insert(1, "yellow")

print(colours)
````

## 7. Removing Items from a List

### `remove()`

The `remove()` method removes the first matching value from a list. It raises a `ValueError` if that value is not present.

**Syntax**
```python
list_name.remove(value)
```

````python
animals = ["cat", "dog", "rabbit", "dog"]
animals.remove("dog")

print(animals)  # Only the first "dog" is removed
````

### `pop()`

The `pop()` method removes and returns an item. If an index is not supplied, it removes the last item.

**Syntax**
```python
list_name.pop()
list_name.pop(index)
```

````python
colours = ["red", "yellow", "blue", "green"]

last_colour = colours.pop()
print("Removed item:", last_colour)
print("List now:", colours)

first_colour = colours.pop(0)
print("Removed first item:", first_colour)
print("List now:", colours)
````

## 8. Arranging Items in a List

### `sort()`

The `sort()` method arranges the items in place. By default, it sorts values in ascending order. Use `reverse=True` for descending order. The items should be comparable with one another.

**Syntax**
```python
list_name.sort()
list_name.sort(reverse=True)
```

````python
numbers = [42, 7, 19, 3, 25]

numbers.sort()
print("Ascending:", numbers)

numbers.sort(reverse=True)
print("Descending:", numbers)
````

### `reverse()`

The `reverse()` method reverses the current order of the list. It does not sort the items; it simply puts the last item first, the second-last item second, and so on.

**Syntax**
```python
list_name.reverse()
```

````python
days = ["Monday", "Tuesday", "Wednesday", "Thursday"]
days.reverse()

print(days)
````

## 9. Counting Items

### `count()`

The `count()` method returns the number of times a particular value appears in a list.

**Syntax**
```python
list_name.count(value)
```

````python
scores = [10, 20, 10, 30, 10, 40]

number_of_tens = scores.count(10)
print("The value 10 appears", number_of_tens, "times.")
````
